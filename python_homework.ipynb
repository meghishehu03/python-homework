```python
# Exercise 12 - Tuples

def old_tuple_hash(t):
    x = 0x345678
    mult = 1000003
    n = len(t)

    for item in t:
        n -= 1
        x = ((x ^ hash(item)) * mult) & ((1 << 64) - 1)
        mult += 82520 + 2 * n

    x = (x + 97531) & ((1 << 64) - 1)

    if x >= (1 << 63):
        x -= (1 << 64)

    return -2 if x == -1 else x


n = int(input())
t = tuple(map(int, input().split()))

print(old_tuple_hash(t))
```

```python
# Exercise 22 - String Formatting
def print_formatted(number):
    width = len(bin(number)[2:])

    for i in range(1, number + 1):
        decimal = str(i)
        octal = oct(i)[2:]
        hexadecimal = hex(i)[2:].upper()
        binary = bin(i)[2:]

        print(decimal.rjust(width),
              octal.rjust(width),
              hexadecimal.rjust(width),
              binary.rjust(width))

if __name__ == '__main__':
    n = int(input())
    print_formatted(n)
```

```python
# Exercise 23 - Alphabet Rangoli
def print_rangoli(size):
    alphabet = "abcdefghijklmnopqrstuvwxyz"
    width = 4 * size - 3
    lines = []

    for i in range(size):
        letters = [alphabet[j] for j in range(size - 1, size - i - 2, -1)]
        pattern = "-".join(letters + letters[-2::-1])
        lines.append(pattern.center(width, "-"))

    for line in lines + lines[-2::-1]:
        print(line)


if __name__ == '__main__':
    n = int(input())
    print_rangoli(n)
```

```python
# Exercise 25 - The Minion Game
def minion_game(string):
    vowels = "AEIOU"
    kevin = 0
    stuart = 0

    for i in range(len(string)):
        if string[i] in vowels:
            kevin += len(string) - i
        else:
            stuart += len(string) - i

    if kevin > stuart:
        print("Kevin", kevin)
    elif stuart > kevin:
        print("Stuart", stuart)
    else:
        print("Draw")


if __name__ == '__main__':
    s = input()
    minion_game(s)
```

```python
# Exercise 47 - Piling Up!
from collections import deque

T = int(input())

for _ in range(T):
    n = int(input())
    cubes = deque(map(int, input().split()))

    last = float('inf')
    possible = True

    while cubes:
        if cubes[0] >= cubes[-1]:
            current = cubes.popleft()
        else:
            current = cubes.pop()

        if current > last:
            possible = False
            break

        last = current

    if possible:
        print("Yes")
    else:
        print("No")
```

```python
# Exercise 49 - Time Delta
#!/bin/python3

import math
import os
import random
import re
import sys

from datetime import datetime
def time_delta(t1, t2):
    formato = "%a %d %b %Y %H:%M:%S %z"
    data1 = datetime.strptime(t1, formato)
    data2 = datetime.strptime(t2, formato)

    differenza = abs((data1 - data2).total_seconds())

    return str(int(differenza))

t = int(input())

for _ in range(t):
    t1 = input()
    t2 = input()

    delta = time_delta(t1, t2)
    print(delta)
```

```python
# Exercise 65 - HTML Parser - Part 1

from html.parser import HTMLParser

class MyHTMLParser(HTMLParser):

    def handle_starttag(self, tag, attrs):
        print("Start :", tag)
        for name, value in attrs:
            print("->", name, ">", value)

    def handle_endtag(self, tag):
        print("End   :", tag)

    def handle_startendtag(self, tag, attrs):
        print("Empty :", tag)
        for name, value in attrs:
            print("->", name, ">", value)

N = int(input())
html = "\n".join(input() for _ in range(N))

parser = MyHTMLParser()
parser.feed(html)
```

```python
# Exercise 66 - HTML Parser - Part 2
from html.parser import HTMLParser

class MyHTMLParser(HTMLParser):

    def handle_comment(self, data):
        if '\n' in data:
            print(">>> Multi-line Comment")
        else:
            print(">>> Single-line Comment")
        print(data)

    def handle_data(self, data):
        if data != '\n':
            print(">>> Data")
            print(data)

html = ""

for _ in range(int(input())):
    html += input().rstrip() + '\n'

parser = MyHTMLParser()
parser.feed(html)
parser.close()
```

```python
# Exercise 69 - Validating Credit Card Numbers
import re

N = int(input())

for _ in range(N):
    numero = input()

    formato = r'[456][0-9]{3}(-?[0-9]{4}){3}'

    if re.fullmatch(formato, numero):
        cifre = numero.replace("-", "")

        if re.search(r'([0-9])\1{3}', cifre):
            print("Invalid")
        else:
            print("Valid")
    else:
        print("Invalid")
```

```python
# Exercise 71 - Matrix Script
#!/bin/python3

import math
import os
import random
import re
import sys

first_multiple_input = input().rstrip().split()

n = int(first_multiple_input[0])
m = int(first_multiple_input[1])
matrix = []

for _ in range(n):
    matrix_item = input()
    matrix.append(matrix_item)

decoded = ''.join(matrix[i][j] for j in range(m) for i in range(n))

result = re.sub(r'(?<=[a-zA-Z0-9])[^a-zA-Z0-9]+(?=[a-zA-Z0-9])', ' ', decoded)

print(result)
```

```python
# Exercise 72 - XML 1 - Find the Score
import xml.etree.ElementTree as etree

def get_attr_number(node):
  return sum(len(element.attrib) for element in node.iter())
n = int(input())
xml = "\n".join(input() for _ in range(n))

root = etree.fromstring(xml)

print(get_attr_number(root))
```

```python
# Exercise 73 - XML2 - Find the Maximum Depth
import xml.etree.ElementTree as etree

maxdepth = 0
def depth(elem, level):
    global maxdepth
    maxdepth = max(maxdepth, level+1)

    for child in elem:
        depth(child, level + 1)

if __name__ == '__main__':
    n = int(input())
    xml = ""
    for i in range(n):
        xml =  xml + input() + "\n"
    tree = etree.ElementTree(etree.fromstring(xml))
    depth(tree.getroot(), -1)
    print(maxdepth)
```

```python
# Exercise 74 - Standardize Mobile Number Using Decorators
def wrapper(f):
    def fun(l):
        numeri = []

        for numero in l:
            numero = numero[-10:]
            numero = "+91 " + numero[:5] + " " + numero[5:]
            numeri.append(numero)

        f(numeri)

    return fun

@wrapper
def sort_phone(l):
    print(*sorted(l), sep='\n')

if __name__ == '__main__':
    l = [input() for _ in range(int(input()))]
    sort_phone(l)
```

```python
# Exercise 75 - Decorators 2 - Name Directory
import operator

def person_lister(f):
    def inner(people):
        return [f(person) for person in sorted(people, key=lambda x: int(x[2]))]
    return inner

@person_lister
def name_format(person):
    return ("Mr. " if person[3] == "M" else "Ms. ") + person[0] + " " + person[1]

if __name__ == '__main__':
    people = [input().split() for i in range(int(input()))]
    print(*name_format(people), sep='\n')
```
